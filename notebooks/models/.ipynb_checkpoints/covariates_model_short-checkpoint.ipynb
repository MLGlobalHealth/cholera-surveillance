## Functions

```python
def custom_smape(y_true, y_pred):
    return np.mean(200* np.abs(y_true - y_pred) / (np.abs(y_true) + np.abs(y_pred)), axis=0)

# def build_fit_xgb(
#     X_train,
#     y_train,
#     learning_rate=0.01,
#     n_estimators=100,
#     max_depth=3,
#     quantiles=None,
#     transform_target=0,
#     random_state=None
# ):
#     if transform_target == 1:
#         likelihood = "quantile"
#         y_train_values = y_train["y"].values().flatten()
#         y_time_index = y_train.time_index
#         y_train = TimeSeries.from_times_and_values(
#             y_time_index,
#             np.log(y_train_values + 1),
#             columns = ["y"]
#         )
#     else:
#         likelihood = "poisson"

#     model = XGBModel(
#         lags_future_covariates=[0],
#         learning_rate=learning_rate,
#         n_estimators=n_estimators,
#         max_depth=max_depth,
#         likelihood=likelihood,
#         quantiles=quantiles,
#         random_state=random_state
#     )
#     model.fit(y_train, future_covariates=X_train)
    
#     return model

def build_fit_xgb(
    X_train,
    y_train,
    learning_rate=0.01,
    n_estimators=100,
    # max_depth=3,
    alpha=0.0001,
    objective="reg:squarederror",
    eval_metric="rmse",
    transform_target=0,
    random_state=None
):
    if transform_target == 1:
        y_train = np.log(y_train + 1)

    model = xgb.XGBRegressor(
        booster = "gblinear",
        learning_rate=learning_rate,
        n_estimators=n_estimators,
        # max_depth=max_depth,
        alpha=alpha,
        objective=objective,
        eval_metric=eval_metric,
        random_state=random_state
    )
    model.fit(X_train, y_train)
    
    return model


# def xgb_objective(trial, X_train, y_train, X_val, y_val, seed=None):
#     # Data Hyperparameters
#     transform_target = trial.suggest_int("transform_target", 0, 1)
    
#     # Fit Hyperparameters
#     quantiles_idx = trial.suggest_int("quantiles", 0, 2)
#     if quantiles_idx == 0:
#         quantiles = None
#     elif quantiles_idx == 1:
#         quantiles = [0.5]
#     else:
#         quantiles = [0.1, 0.5, 0.9]

#     # XGBoost Hyperparameters
#     learning_rate = trial.suggest_float("learning_rate", 0.005, 0.1, log=True)
#     n_estimators = trial.suggest_int("n_estimators", 100, 1000)
#     max_depth = trial.suggest_int("max_depth", 3, 10)
    
#     # Model
#     fit = build_fit_xgb(
#         X_train = X_train,
#         y_train = y_train,
#         learning_rate=learning_rate,
#         n_estimators=n_estimators,
#         max_depth=max_depth,
#         transform_target=transform_target,
#         quantiles=quantiles,
#         random_state=seed
#     )
    
#     # Predict
#     y_val_pred = fit.predict(
#         n=X_val.shape[0],
#         future_covariates = X_val,
#     )
#     if transform_target == 1:
#         y_val_pred_values = y_val_pred["y"].values().flatten()
#         y_val_time_index = y_val_pred.time_index
#         y_val_pred = TimeSeries.from_times_and_values(
#             y_val_time_index,
#             np.log(y_val_pred_values + 1),
#             columns = ["y"]
#         )
#     smape_all = smape(y_val, y_val_pred)
#     smape_mean = np.mean(smape_all)
#     return smape_mean if smape_mean == smape_mean else np.inf


def xgb_objective(trial, X_train, y_train, X_val, y_val, seed=None):
    # Data Hyperparameters
    transform_target = trial.suggest_int("transform_target", 0, 1)

    # XGBoost Hyperparameters
    learning_rate = trial.suggest_float("learning_rate", 0.005, 0.1, log=True)
    n_estimators = trial.suggest_int("n_estimators", 100, 1000)
    # max_depth = trial.suggest_int("max_depth", 3, 10)
    alpha = trial.suggest_float("alpha", 0.001, 1000, log=True)
    objective = trial.suggest_categorical("objective", ["reg:squarederror", "reg:pseudohubererror", "reg:absoluteerror"])
    eval_metric = trial.suggest_categorical("eval_metric", ["rmse", "mae", "rmsle", "mape", "logloss", "poisson-nloglik", "gamma-nloglik"])
    
    # Model
    fit = build_fit_xgb(
        X_train = X_train,
        y_train = y_train,
        learning_rate=learning_rate,
        n_estimators=n_estimators,
        # max_depth=max_depth,
        alpha=alpha,
        objective=objective,
        eval_metric=eval_metric,
        transform_target=transform_target,
        random_state=seed
    )
    
    # Predict
    y_val_pred = fit.predict(X_val)
    if transform_target == 1:
        y_val_pred = np.exp(y_val_pred) - 1
    smape_all = custom_smape(y_val, y_val_pred)
    smape_mean = np.mean(smape_all)
    return smape_mean if smape_mean == smape_mean else np.inf


def build_fit_lasso(
    X_train,
    y_train,
    alpha=0.0001,
    max_iter=1000,
    tol=1e-5,
    transform_target=0,
    random_state=None
):
    if transform_target == 1:
        y_train = np.log(y_train + 1)

    model = Lasso(
        alpha=alpha,
        max_iter=max_iter,
        tol=tol,
        random_state=random_state
    )
    model.fit(X_train, y_train)
    
    return model


def lasso_objective(trial, X_train, y_train, X_val, y_val, seed=None): 
    # Data Hyperparameters
    transform_target = trial.suggest_int("transform_target", 0, 1)
    
    # Lasso Hyperparameters
    alpha = trial.suggest_float("alpha", 0.001, 1000, log=True)
    max_iter = trial.suggest_int("max_iter", 1000, 10000)
    tol = trial.suggest_float("tol", 1e-5, 1e-2, log=True)
    # Model
    fit = build_fit_lasso(
        X_train = X_train,
        y_train = y_train,
        alpha=alpha,
        max_iter=max_iter,
        tol=tol,
        transform_target=transform_target,
        random_state=seed
    )
    # Predict
    y_val_pred = fit.predict(X_val)
    if transform_target == 1:
        y_val_pred = np.exp(y_val_pred) - 1
    smape_all = custom_smape(y_val, y_val_pred)
    smape_mean = np.mean(smape_all)
    return smape_mean if smape_mean == smape_mean else np.inf




def xgbr_config(trial: optuna.Trial):
    return {
        "verbosity": 0,
        "learning_rate": trial.suggest_float("learning_rate", 0.005, 0.1, log=True),
        "n_estimators": trial.suggest_int("n_estimators", 100, 1000),
        "max_depth": trial.suggest_int("max_depth", 3, 10),        
    }

def lasso_config(trial: optuna.Trial):
    return {
        "alpha": trial.suggest_float("alpha", 0.0001, 1, log=True),
        "max_iter": trial.suggest_int("max_iter", 1000, 10000),
        "tol": trial.suggest_float("tol", 1e-5, 1e-2, log=True),
    }

def ridge_config(trial: optuna.Trial):
    return {
        "alpha": trial.suggest_float("alpha", 0.0001, 1, log=True),
        "max_iter": trial.suggest_int("max_iter", 1000, 10000),
        "tol": trial.suggest_float("tol", 1e-5, 1e-2, log=True),
    }

def my_init_config(trial: optuna.Trial):
    lag_transforms_options = [
        ExpandingMean(),
        ExponentiallyWeightedMean(alpha=0.3),
        RollingMean(
            window_size=int(28 / TIME_BIN), 
            min_samples=1
        ),
    ]
    n_lags = trial.suggest_int("n_lags", 0, 8)
    lags = rng.choice(range(1, 9), n_lags, replace=False).tolist()
    n_lag_transforms = trial.suggest_int("n_transforms", 0, 3)
    lag_trans = rng.choice(lag_transforms_options, n_lag_transforms, replace=False).tolist()
    lag_transforms = {}
    for lag in lags:
        lag_transforms[lag] = lag_trans

    target_transforms_options = [
        LocalBoxCox(),
        Differences([1, 52]),
    ]
    n_target_transforms = trial.suggest_int("n_target_transforms", 0, 2)
    if n_target_transforms == 0:
        target_transforms = []
    elif n_target_transforms == 1:
        target_transforms = rng.choice(target_transforms_options, 1).tolist()
    else:
        target_transforms = target_transforms_options

    return {
        "lags": lags,  
        "lag_transforms": lag_transforms,
        "target_transforms": target_transforms,
    }

def print_callback(study, trial):
    print(f"Current value: {trial.value}, Current params: {trial.params}")
    print(f"Best value: {study.best_value}, Best params: {study.best_trial.params}")


def init_config_simple(trial: optuna.Trial):
    return {}

# Aggregation functions
def na_sum(x):
    if all(x.isna()):
        return np.nan
    else:
        return x.dropna().sum()

def na_log_sum(x):
    if all(x.isna()):
        return np.nan
    else:
        tot = x.dropna().sum()
        if tot > 0:
            return np.log(tot + 1)
        elif tot == 0:
            return 0
        else:
            return -np.log(-tot + 1)

def na_mean(x):
    if all(x.isna()):
        return np.nan
    else:
        return x.dropna().mean()


def standardize(x):
    return (x - x.mean()) / x.std()


def normalize(x):
    return (x - x.min()) / (x.max() - x.min())


def mask_target(x):
    """
    this mask returns an array of 1s where the last entry is a 0.
    Used to mask the rolled time series to avoid using the target in the features.
    """
    result = np.ones(len(x))
    result[-1] = 0
    return result
```

```python
df = pd.read_csv(
    processed_data_dir / "uvira_full_data.csv",
    parse_dates=["date"]
)
```

## Parameters

```python
START_DATE = "2016-05-01"
TRAIN_DATE = "2019-01-01"
VAL_DATE = "2020-01-01"
TIME_BIN = 7
FORECAST_HORIZON = int(30 / TIME_BIN)
```

## Aggregate Data

```python
df_weekly = (
    df
    .groupby(pd.Grouper(key = "date", freq = f"{TIME_BIN}D"), dropna=True)
    .agg({
        "reports": na_sum,
        "rdt_confirmed": na_sum,
        "vacc_1": na_sum,
        "vacc_1_prop": na_sum,
        "vacc_2": na_sum,
        "vacc_2_prop": na_sum,
        "rainfall_mm": na_mean,
        "esi": na_mean,
        "lake_level": na_mean,
        "spei": na_mean,
        "num_fatalities": na_log_sum,
        "num_disaster_affected": na_sum,
        "idps": na_sum,
    })
    .assign(
        idps = lambda x: x["idps"].fillna(0)
    )
    .loc[lambda x: (
        (x.index >= START_DATE)
    )]
    .reset_index()
    .rename(columns = {
        "date": "ds",
        "reports": "y"
    })
)
```

```python
df_weekly_for_model = (
    df_weekly
    .assign(
        unique_id = 1,
    )
    [[
        "unique_id",
        "ds",
        "y",
        "rainfall_mm",
        "esi",
        "lake_level",
        "spei",
        "num_fatalities",
        "num_disaster_affected",
    ]]
    .dropna()
    .reset_index(drop = True)
)
```

```python
(
    df_weekly_for_model
    .drop(columns = ["unique_id"])
    .set_index("ds")
    .plot(subplots = True, figsize = (6, 6))
)
```

## Feature Engineering

### Create rolling datasets

Note that the rolling procedure includes the current timestamp so we need to drop the current value

```python
lookback_period = 4
df_weekly_rolled = (
    roll_time_series(
        df_weekly_for_model.drop(columns=["y"]),
        column_id = "unique_id",
        column_sort = "ds",
        min_timeshift = lookback_period,
        max_timeshift = lookback_period
    )
    .drop(columns = ["unique_id"])
)
```

```python
mask = (
    df_weekly_rolled
    .groupby("id")
    ["id"]
    .transform(mask_target)
    .astype(bool)
    .values
)
```

```python
df_weekly_rolled_masked = (
    df_weekly_rolled
    .copy()
    [mask]
    .reset_index(drop = True)
)
```

### Build features

```python
df_features = (
    extract_features(
        df_weekly_rolled_masked,
        column_id = "id",
        column_sort = "ds",
    )
    [lambda x: [c for c in x.columns if (x[c].isna().sum() == 0) & (x[c].std() > 0)]]
)
```

### Clean features

```python
df_ts_model_features = (
    impute(df_features)
    .reset_index()
    .drop(columns = ["level_0"])
    .rename(columns = {"level_1": "ds"})
    .dropna(axis = 1)
)
```

## Data for model

We use all the engineered features and add a few more manually. Then we'll let the models figure out which ones are most important.

```python
df_all_data = (
    df_weekly
    .assign(
        week = lambda x: 0.5 * (1 + np.sin(2 * np.pi * (x["ds"].dt.isocalendar().week + 30) / 52)),
        rainfall_mm_sum_4w = lambda x: (sum([x["rainfall_mm"].shift(i) for i in range(1, 5)])),
        rainfall_mm_sum_8w = lambda x: (sum([x["rainfall_mm"].shift(i) for i in range(1, 9)])),
        esi_mean_4w = lambda x: (sum([x["esi"].shift(i) for i in range(1, 5)])/4),
        esi_mean_8w = lambda x: (sum([x["esi"].shift(i) for i in range(1, 9)])/8),
        lake_level_mean_4w = lambda x: (sum([x["lake_level"].shift(i) for i in range(1, 5)])/4),
        lake_level_mean_8w = lambda x: (sum([x["lake_level"].shift(i) for i in range(1, 9)])/8),
        spei_mean_4w = lambda x: (sum([x["spei"].shift(i) for i in range(1, 5)])/4),
        spei_mean_8w = lambda x: (sum([x["spei"].shift(i) for i in range(1, 9)])/8),
        num_fatalities_sum_4w = lambda x: (sum([x["num_fatalities"].shift(i) for i in range(1, 5)])),
        num_fatalities_sum_8w = lambda x: (sum([x["num_fatalities"].shift(i) for i in range(1, 9)])),
        num_disaster_affected_sum_4w = lambda x: (sum([x["num_disaster_affected"].shift(i) for i in range(1, 5)])),
        num_disaster_affected_sum_8w = lambda x: (sum([x["num_disaster_affected"].shift(i) for i in range(1, 9)])),
    )
    .pipe(
        lambda x: x.assign(
            **{
                f"{col}_{i}w": lambda x, col=col, i=i: x[col].shift(i) 
                for i in range(1, 9)
                for col in ["rainfall_mm", "esi", "lake_level", "spei", "num_fatalities", "num_disaster_affected"]
            }
        )
    )
    .drop(columns = [c for c in df_weekly.columns if c not in ["ds", "y", "unique_id"]])
    .merge(
        df_ts_model_features,
        how = "left",
        on = "ds"
    )
    .dropna()
    .reset_index(drop = True)
    .set_index("ds")
)
    
```

```python
df_all_data = (
    df_all_data
    .pipe(
        lambda x: x.assign(
            **{
                num_col: lambda df_, num_col=num_col: df_[num_col].astype("float32") for num_col in x.select_dtypes(include = ["float64"]).columns
            }
        )
    )
)
```

## Train-validation-test split

```python
X_train = df_all_data.loc[lambda x: x.index < TRAIN_DATE].drop(columns = ["y"])
y_train = df_all_data.loc[lambda x: x.index < TRAIN_DATE]["y"]
X_val = df_all_data.loc[lambda x: (x.index >= TRAIN_DATE) & (x.index < VAL_DATE)].drop(columns = ["y"])
y_val = df_all_data.loc[lambda x: (x.index >= TRAIN_DATE) & (x.index < VAL_DATE)]["y"]
X_test = df_all_data.loc[lambda x: x.index >= VAL_DATE].drop(columns = ["y"])
y_test = df_all_data.loc[lambda x: x.index >= VAL_DATE]["y"]
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(y_train.index, y_train, label="Train")
ax.plot(y_val.index, y_val, label="Validation")
ax.plot(y_test.index, y_test, label="Test")
ax.legend(fontsize=12)
ax.set_title("Dataset Splits", fontsize=16)
ax.tick_params(labelsize = 12)
ax.set_title("Weekly Suspected Cases", fontsize = 16)
ax.set_xlabel("Date", fontsize = 14)
ax.set_ylabel("Count", fontsize = 14)
ax.set(xticks=["2016-01-01", "2018-01-01", "2020-01-01", "2022-01-01", "2024-01-01"])
ax.set_xticklabels(["2016", "2018", "2020", "2022", "2024"], fontsize = 12) 
plt.tight_layout()
plt.minorticks_off()
fig.align_ylabels()
plt.savefig(img_dir / "train_test_split.png", dpi = 300, bbox_inches = "tight")
```

## Data Preprocessing

```python
max_abs_scaler = MaxAbsScaler()
X_train_scaled = max_abs_scaler.fit_transform(X_train)
X_val_scaled = max_abs_scaler.transform(X_val)
X_test_scaled = max_abs_scaler.transform(X_test)

X_train_ts = TimeSeries.from_times_and_values(
    times = X_train.index,
    values = X_train_scaled,
    freq = "7D",
    columns = X_train.columns
)
y_train_ts = TimeSeries.from_times_and_values(
    times = y_train.index,
    values = y_train.values,
    freq = "7D",
    columns = ["y"]
)
X_val_ts = TimeSeries.from_times_and_values(
    times = X_val.index,
    values = X_val_scaled,
    freq = "7D",
    columns = X_val.columns
)
y_val_ts = TimeSeries.from_times_and_values(
    times = y_val.index,
    values = y_val.values,
    freq = "7D",
    columns = ["y"]
)
X_test_ts = TimeSeries.from_times_and_values(
    times = X_test.index,
    values = X_test_scaled,
    freq = "7D",
    columns = X_test.columns
)
y_test_ts = TimeSeries.from_times_and_values(
    times = y_test.index,
    values = y_test.values,
    freq = "7D",
    columns = ["y"]
)

df_train = (
    pd.DataFrame(
        X_train_scaled, 
        columns = X_train.columns,
        index = X_train.index
    )
    .assign(
        y = y_train.values,
    )
    .reset_index()
)
df_val = (
    pd.DataFrame(
        X_val_scaled, 
        columns = X_val.columns,
        index = X_val.index
    )
    .assign(
        y = y_val.values,
    )
    .reset_index()
)
df_test = (
    pd.DataFrame(
        X_test_scaled, 
        columns = X_test.columns,
        index = X_test.index
    )
    .assign(
        y = y_test.values,
    )
    .reset_index()
)
```

## Simple Correlation Analysis

```python
corr_factors = (
    df_train
    .drop(columns = ["ds"])
    .corr()
    ["y"]
    .T
    .to_frame(name = "corr")
    .loc[lambda x: x.index != "y"]
    .loc[lambda x: x["corr"] >= 0.01]
    .sort_values(by = "corr", ascending = False, key = np.abs)
)
```

#### Extract features with low correlation

```python
list_of_relevant_features_simple = [corr_factors.index.tolist()[0]]
for f in corr_factors.index.tolist()[1:]:
    corrs = df_train[list_of_relevant_features_simple + [f]].corr()[f].abs()
    if corrs[:-1].max() < 0.2:
        list_of_relevant_features_simple.append(f)    
```

```python
len(list_of_relevant_features_simple)
```

```python
corr_factors
```

#### Final Dataset

```python
df_reports_cov_corr_simple = (
    df_test
    .reset_index()
    [["ds", "y"] + list_of_relevant_features_simple]
    .assign(
        **{
            f"{feat}_corr": lambda x, feat=feat: x[feat] * corr_factors.loc[lambda x: x.index == feat, "corr"].values[0]
            for feat in list_of_relevant_features_simple
        }
    )
    .assign(
        cov = lambda x: np.exp(normalize(
            x
            [[c for c in x.columns if c.endswith("_corr")]]
            .sum(axis = 1)
        )),
        # y = lambda x: np.exp(x["y"]) - 1,
    )
    .merge(
        df_weekly[["ds", "idps", "num_disaster_affected"]],
        on = "ds",
        how = "left"
    )
    .rename(
        columns = {
            "ds": "date",
            "y": "reports"
        }
    )
    [["date", "reports", "idps", "num_disaster_affected", "cov"]]
)
    
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
(
    df_reports_cov_corr_simple
    .set_index("date")
    .plot(
        y = "cov",
        ax = ax, 
        legend = False
    )
)
ax2 = ax.twinx()
(
    df_reports_cov_corr_simple
    .set_index("date")
    .plot(
        y = "reports",
        ax = ax2, 
        legend = False,
        color = "firebrick"
    )
)
ax.set_ylabel("Combined covariate\n(norm.)", fontsize = 14)
ax.set(xticks=["2020-01-01", "2021-01-01", "2022-01-01", "2023-01-01", "2024-01-01"])
ax.set_xticklabels(["2020", "2021", "2022", "2023", "2024"], fontsize = 12) 
ax.set_xlabel("Date", fontsize = 14)
ax.tick_params(axis = "y", labelsize = 12)
plt.minorticks_off()
plt.savefig(img_dir / "combined_covariates.png", dpi = 300, bbox_inches = "tight")
```

```python
# Data point to plot
data_to_plot = xgbr_shap_values_train
p = 1

# Extract feature names, shap values, and data values
feature_names = data_to_plot[p].feature_names
shap_values = data_to_plot[p].values
data_values = data_to_plot[p].data

# Get the absolute values of the SHAP values and sort them
sorted_indices = np.argsort(np.abs(shap_values))[::-1]
sorted_feature_names = np.array(feature_names)[sorted_indices]
sorted_shap_values = shap_values[sorted_indices]
sorted_data_values = data_values[sorted_indices]

# Select the top features
top_n = 8
top_feature_names = sorted_feature_names[:top_n][::-1]
top_feature_names = [n[:20] + "..." if len(n) > 20 else n for n in top_feature_names]
labels = [f"{name} ({value:.2f})" for name, value in zip(top_feature_names, sorted_data_values[:top_n][::-1])]
top_shap_values = sorted_shap_values[:top_n][::-1]

# Calculate the cumulative sum of the SHAP values
cumulative_shap_values = -(np.cumsum(top_shap_values) - np.cumsum(top_shap_values)[-1])

# Create the waterfall plot with arrows
fig, ax = plt.subplots(figsize=(6, 4))
colors = ["crimson" if val < 0 else "deepskyblue" for val in top_shap_values]
for i in range(top_n):
    ax.arrow(
        cumulative_shap_values[i], 
        i, 
        top_shap_values[i], 
        0,
        head_width=0.1, 
        head_length=0.1, 
        length_includes_head = True,
        fc=colors[i], 
        ec=colors[i],
        linewidth = 10
    )
ax.set_yticks(range(top_n))
ax.set_yticklabels(labels)
ax.set_xlabel("SHAP value", fontsize=14)
ax.set_ylabel("Feature", fontsize=14)
ax.tick_params(labelsize=12)
ax.set_title("Top Features - Single Date", fontsize=16)
plt.savefig(img_dir / "shap_waterfall.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

```python
# Data point to plot
p = 1

# Extract feature names, shap values, and data values
feature_names = xgbr_shap_values_all[p].feature_names
shap_values = xgbr_shap_values_all[p].values
data_values = xgbr_shap_values_all[p].data

# Get the absolute values of the SHAP values and sort them
sorted_indices = np.argsort(np.abs(shap_values))[::-1]
sorted_feature_names = np.array(feature_names)[sorted_indices]
sorted_shap_values = shap_values[sorted_indices]
sorted_data_values = data_values[sorted_indices]

# Select the top features
top_n = 8
top_feature_names = sorted_feature_names[:top_n][::-1]
top_feature_names = [n[:20] + "..." if len(n) > 20 else n for n in top_feature_names]
labels = [f"{name} ({value:.2f})" for name, value in zip(top_feature_names, sorted_data_values[:top_n][::-1])]
top_shap_values = sorted_shap_values[:top_n][::-1]

# Calculate the cumulative sum of the SHAP values
cumulative_shap_values = -(np.cumsum(top_shap_values) - np.cumsum(top_shap_values)[-1])

# Create the waterfall plot with arrows
fig, ax = plt.subplots(figsize=(6, 4))
colors = ["crimson" if val < 0 else "deepskyblue" for val in top_shap_values]
for i in range(top_n):
    ax.arrow(
        cumulative_shap_values[i], 
        i, 
        top_shap_values[i], 
        0,
        head_width=0.1, 
        head_length=0.1, 
        length_includes_head = True,
        fc=colors[i], 
        ec=colors[i],
        linewidth = 10
    )
ax.set_yticks(range(top_n))
ax.set_yticklabels(labels)
ax.set_xlabel("SHAP value", fontsize=14)
ax.set_ylabel("Feature", fontsize=14)
ax.tick_params(labelsize=12)
ax.set_title("Top Features - Single Date", fontsize=16)
plt.savefig(img_dir / "shap_waterfall.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

#### Combine SHAP values

This requires a bit of thinking.
A negative SHAP value is associated to a reduction in the output regardless of the sign of the covariate so we cannot multiply the two without losing this information.

```python
(
    df_model
    [["ds"] + list_of_relevant_features_all]
    .assign(
        **{
            f"{feat}_shap": np.sign(df_xgbr_shap_all[feat]) * df_model[feat] * df_xgbr_shap_importances_all.loc[lambda x: x["feature"] == feat]["importance"].values[0]
            for feat in list_of_relevant_features_all
        }
    )
    .drop(columns = [c for c in list_of_relevant_features_all])
    .set_index("ds")
    .plot(legend=False)
)
```

#### Final Dataset

```python
df_reports_cov_shap_all = (
    df_model
    [["ds", "y"] + list_of_relevant_features_all]
    .assign(
        **{
            f"{feat}_shap": np.sign(df_xgbr_shap_all[feat]) * df_model[feat] * df_xgbr_shap_importances_all.loc[lambda x: x["feature"] == feat]["importance"].values[0]
            for feat in list_of_relevant_features_all
        }
    )
    .assign(
        cov = lambda x: normalize(
            x
            [[c for c in x.columns if c.endswith("_shap")]]
            .sum(axis = 1)
        )
    )
    .merge(
        df_weekly[["ds", "rdt_confirmed", "idps", "num_disaster_affected", "vacc_1", "vacc_1_prop", "vacc_2", "vacc_2_prop"]],
        on = "ds",
        how = "left"
    )
    .rename(
        columns = {
            "ds": "date",
            "y": "reports"
        }
    )
    [["date", "reports", "rdt_confirmed", "idps", "num_disaster_affected", "vacc_1", "vacc_1_prop", "vacc_2", "vacc_2_prop", "cov"] + list_of_relevant_features_all]
)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
(
    df_reports_cov_shap_all
    .set_index("date")
    .plot(
        y = "cov",
        ax = ax, 
        legend = False
    )
)
ax2 = ax.twinx()
(
    df_reports_cov_shap_all
    .set_index("date")
    .plot(
        y = "reports",
        ax = ax2, 
        legend = False,
        color = "firebrick"
    )
)
ax.set_ylabel("Combined covariate\n(norm.)", fontsize = 14)
ax.set(xticks=["2016-01-01", "2018-01-01", "2020-01-01", "2022-01-01", "2024-01-01"])
ax.set_xticklabels(["2016", "2018", "2020", "2022", "2024"], fontsize = 12) 
ax.set_xlabel("Date", fontsize = 14)
ax.tick_params(axis = "y", labelsize = 12)
fig.align_ylabels(axes)
plt.minorticks_off()
plt.savefig(img_dir / "combined_covariates.png", dpi = 300, bbox_inches = "tight")
```

### Use correlation factors to build covariance

```python
corr_factors_all = (
    df_model
    [["y"] + list_of_relevant_features_all]
    .corr()
    ["y"]
    .T
    .to_frame(name = "corr")
    .loc[lambda x: x.index != "y"]
    .sort_values(by = "corr", ascending = False, key = np.abs)
)
```

#### Final Dataset

```python
df_reports_cov_corr_all = (
    df_model
    [["ds", "y"] + list_of_relevant_features_all]
    .assign(
        **{
            f"{feat}_corr": df_model[feat] * corr_factors_all.loc[lambda x: x.index == feat, "corr"].values[0]
            for feat in list_of_relevant_features_all
        }
    )
    .assign(
        cov = lambda x: normalize(
            x
            [[c for c in x.columns if c.endswith("_corr")]]
            .sum(axis = 1)
        )
    )
    .merge(
        df_weekly[["ds", "rdt_confirmed", "idps", "num_disaster_affected", "vacc_1", "vacc_1_prop", "vacc_2", "vacc_2_prop"]],
        on = "ds",
        how = "left"
    )
    .rename(
        columns = {
            "ds": "date",
            "y": "reports"
        }
    )
    [["date", "reports", "rdt_confirmed", "idps", "num_disaster_affected", "vacc_1", "vacc_1_prop", "vacc_2", "vacc_2_prop", "cov"] + list_of_relevant_features_all]
)
    
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
(
    df_reports_cov_corr_all
    .set_index("date")
    .plot(
        y = "cov",
        ax = ax, 
        legend = False
    )
)
ax2 = ax.twinx()
(
    df_reports_cov_corr_all
    .set_index("date")
    .plot(
        y = "reports",
        ax = ax2, 
        legend = False,
        color = "firebrick"
    )
)
ax.set_ylabel("Combined covariate\n(norm.)", fontsize = 14)
ax.set(xticks=["2016-01-01", "2018-01-01", "2020-01-01", "2022-01-01", "2024-01-01"])
ax.set_xticklabels(["2016", "2018", "2020", "2022", "2024"], fontsize = 12) 
ax.set_xlabel("Date", fontsize = 14)
ax.tick_params(axis = "y", labelsize = 12)
fig.align_ylabels(axes)
plt.minorticks_off()
plt.savefig(img_dir / "combined_covariates.png", dpi = 300, bbox_inches = "tight")
```
